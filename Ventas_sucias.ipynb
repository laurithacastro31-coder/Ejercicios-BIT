2.LIMPIEZA DE DATOS PANDAS


```python
# manejo de nulos
vc1["precio"] = vc1["precio"].fillna("desconocido")
vc1["cliente"] = vc1["cliente"].fillna("desconocido")
vc1["producto"] = vc1["producto"].fillna("desconocido")
vc1["cantidad"] = vc1["cantidad"].fillna("desconocido")
vc1["pais"] = vc1["pais"].fillna("desconocido")
vc1["metodo_pago"] = vc1["metodo_pago"].fillna("desconocido")
vc1["fecha"] = vc1["fecha"].fillna("desconocido")
```

```python
#Convertir vacios
vc1 ["precio"] = vc1["precio"].replace("",None)
vc1 ["precio"] = vc1["precio"].replace(" ",None)

vc1 ["cliente"] = vc1["cliente"].replace("",None)
vc1 ["cliente"] = vc1["cliente"].replace(" ",None)

vc1 ["producto"] = vc1["producto"].replace("",None)
vc1 ["producto"] = vc1["producto"].replace(" ",None)

vc1 ["cantidad"] = vc1["cantidad"].replace("",None)
vc1 ["cantidad"] = vc1["cantidad"].replace(" ",None)

vc1 ["pais"] = vc1["pais"].replace("",None)
vc1 ["pais"] = vc1["pais"].replace(" ",None)

vc1 ["metodo_pago"] = vc1["metodo_pago"].replace("",None)
vc1 ["metodo_pago"] = vc1["metodo_pago"].replace(" ",None)

vc1 ["fecha"] = vc1["fecha"].replace("",None)
vc1 ["fecha"] = vc1["fecha"].replace(" ",None)
```

```python
#Mapeo
vc1["pais"] = vc1["pais"].replace({
    "COL":"colombia",
    "col": "colombia",
    "Colombia" :"colombia",
    "Chile": "chile",
    "Perú" : "peru"
    })
```

```python
#formato de fecha
vc1['fecha'] = pd.to_datetime(vc1['fecha'], errors='coerce', format='mixed', dayfirst=True)
```

```python
#convertir la columna a numerica
vc1["precio"]=pd.to_numeric(vc1["precio"], errors="coerce")
```

```python
vc1["precio"].dtype
```

```python
# Manejo de outliers

vc1 = vc1[vc1["precio"]<100000]
```

```python
vc1.head()
```

en esta parte se realizo limpieza de datos corrigiendo vacios, estandarizando formato de fecha y dejando valores consistentes

4.INTRODUCCION A NUMPY

```python
#Convertir columnas numéricas a un array
data = vc1[['precio', 'cantidad','Total']].to_numpy()
```

```python
#separar columnas
precios = data[:, 0]
cantidades = data[:, 1]
Total= data[:,2]
```

```python
#vectorizar para calcular las ventas
ventas_totales = precios * cantidades
print(ventas_totales)

```

5.ANALISIS CON NUMPY

```python
#suma total de ventas
total_ventas=np.sum(ventas_totales)
print(total_ventas)
```

```python
#promedio de ventas
promedio_ventas=np.mean(ventas_totales)
print(promedio_ventas)
```

```python
#venta maxima
venta_maxima=np.max(ventas_totales)
print(venta_maxima)
```

```python
#venta minima
venta_minima=np.min(ventas_totales)
print(venta_minima)
```

```python
#cantidad de venta superiores a 1000
ventas_mayores_1000=np.sum(ventas_totales>1000)
print(ventas_mayores_1000)
```
