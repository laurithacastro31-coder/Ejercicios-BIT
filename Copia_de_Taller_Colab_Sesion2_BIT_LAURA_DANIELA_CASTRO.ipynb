## Ejercicio 9. Excepciones básicas
**Tiempo sugerido:** 5 minutos

**Objetivo:** Evitar que el programa falle ante un dato incorrecto.

**Instrucciones**

3. Si el usuario escribe algo no numérico, captura el error con except ValueError e imprime un mensaje claro.

**Pista:** La estructura es try: ... except ValueError: ...


```python
usuario = input("introduce un numero")
try:
    numero = int(usuario)
    print("Conversión exitosa:", numero)
except ValueError:
    print("Error: No se pudo convertir a entero.")


```
