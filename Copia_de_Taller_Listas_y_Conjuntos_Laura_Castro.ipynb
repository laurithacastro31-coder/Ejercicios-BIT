Ejercicio 1: Clientes Duplicados y Segmentación

Contexto:

Tienes una lista de clientes con datos repetidos. Además, cuentas con dos listas: clientes
activos e inactivos.

Objetivos:
1. Eliminar duplicados de la lista original.
2. Identificar clientes en ambos estados.
3. Encontrar clientes solo activos.
4. Recorrer los clientes y clasificarlos.

Pistas:

1. Usa set() para eliminar duplicados.
2. Usa intersection() para encontrar coincidencias.
3. Usa difference() para encontrar diferencias.
4. Usa un for con if para clasificar.

```python
clientes_activos = ["Jorge", "Laura", "Angie", "Martin", "Samuel", "Helena", "Sofia", "David", "Sofia", "Laura", "Jorge"]
clientes_inactivos = ["David", "Vivian", "Samuel", "Pablo", "Sofia", "Andres", "Laura", "Jorge", "Valentina", "Lucas"]

# Convertimos las listas a set para eliminar los duplicados.
lista_clientes_activos = set(clientes_activos)
lista_clientes_inactivos = set(clientes_inactivos)

# Encontramos a los clientes que estan en ambas listas.
clientes_ambas_listas = lista_clientes_activos.intersection(lista_clientes_inactivos)

# usamos difference para ubicar los clientes que solo estan en una lista.
clientes_solo_activos = lista_clientes_activos.difference(lista_clientes_inactivos)
clientes_solo_inactivos = lista_clientes_inactivos.difference(lista_clientes_activos)

# Luego de investigar, usare el metodo union para mostrar todos los clientes sin importar su estado.
todos_los_clientes = lista_clientes_activos.union(lista_clientes_inactivos)

# Usa un for con if para clasificar
alert = "ℹ️"
inactive = "🔴"
active = "🟢"

print("Informe de clientes:\n")

for cliente in todos_los_clientes:
    if cliente in clientes_ambas_listas:
        print(f"{alert} {cliente} está en ambas listas. (Revisar)")
    elif cliente in clientes_solo_activos:
        print(f"{active} {cliente} está activo.")
    elif cliente in clientes_solo_inactivos:
        print(f"{inactive} {cliente} está inactivo.")


#for cliente in clientes_activos:
#   if cliente in clientes_inactivos:
#       print(f"{cliente} está en ambas listas.")
#   else:
#        print(f"{cliente}  está activo.")
#
# Este bucle me muestra clientes activos y en ambas listas. Investigando como evitar los duplicados en el resultado.

```

Ejercicio 2: Productos y Análisis de Inventario

Contexto:
Tienes una lista de productos vendidos y otra de inventario disponible.

Objetivos:

1. Identificar todos los productos.
2. Encontrar productos vendidos que están en inventario.
3. Detectar productos que no se han vendido.
4. Evaluar productos con una condición usando while.


Pistas:

1. Usa union() para combinar conjuntos.
2. Usa & para intersección.
3. Usa - para diferencia.
4. Convierte el set a lista para usar while.



```python
producto_vendidos = ["Mouse", "Escritorio", "Telefono", "Impresora"]
inventario_disponible = ["Escritorio", "Teclado", "Monitor", "Telefono", "Mouse", "Auriculares", "Laptop"]

#Combinamos todos los productos.
todos_los_productos = set(producto_vendidos).union(set(inventario_disponible))

# Verificar productos vendidos que aun siguen disponibles en el inventario.
productos_vendidos_en_inventario = set(producto_vendidos) & (set(inventario_disponible))

#Ubicamos productos no vendidos que estan disponibles en el inventario.
productos_no_vendidos = set(inventario_disponible).difference(set(producto_vendidos))


#evaluamos productos con una condicion usando while
evaluacion = list(todos_los_productos)
i = 0

while i < len(evaluacion):
    producto = evaluacion[i]
    if producto in productos_vendidos_en_inventario:
        print(f"💲 {producto}  vendido y en stock.")
    elif producto in productos_no_vendidos:
        print(f"📦 {producto} No se ha vendido y sigue disponible. ")
    i += 1




```

Ejercicio 3: Estudiantes y Cursos.

Contexto:

Tienes estudiantes inscritos en dos cursos distintos.

Objetivos:

1. Encontrar estudiantes en ambos cursos.
2. Identificar estudiantes exclusivos por curso.
3. Unir todos los estudiantes.
4. Clasificar estudiantes según su participación.

Pistas:

1. Usa & para intersección.
2. Usa difference() para exclusivos.
3. Usa union() para todos.
4. Usa for + if para clasificar.

```python
curso_ux = ["David", "Laura", "Vivian", "Martin", "Isis", "Gabriel", "Helena", "Sofia"]
curso_python = ["David", "Vivian", "Pedro", "Pablo", "Sofia", "Camila", "Diego", "Isis"]

star = "⭐"
ux = "🎨"
snake= "🐍"



# Encontramos estudiantes que estan en ambos cursos.
estudiantes_ambos_cursos = set(curso_ux) & set(curso_python)

# Encontramos estudiantes que solo estan en un curso.
estudiantes_solo_ux = set(curso_ux) - set(curso_python)
estudiantes_solo_python = set(curso_python) - set(curso_ux)

# Combinamos todos los estudiantes.
todos_los_estudiantes = set(curso_ux).union(set(curso_python))

# clasificamos a los estudiantes usando un for con if segun su participacion en los cursos.
print("\nInforme de estudiantes:\n")
for estudiante in todos_los_estudiantes:
    if estudiante in estudiantes_ambos_cursos:
        print(f"{star} {estudiante} está en ambos cursos. ")
    elif estudiante in estudiantes_solo_ux:
        print(f"{ux} {estudiante} está solo en el curso de UX.")
    elif estudiante in estudiantes_solo_python:
        print(f"{snake} {estudiante} está solo en el curso de Python.")
```
