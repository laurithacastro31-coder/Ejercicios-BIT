## 1. Recordatorio: lista vs conjunto

### Lista
Una lista se usa cuando:
- importa el orden,
- puede haber datos repetidos,
- quieres acceder por posición.

### Conjunto (`set`)
Un conjunto se usa cuando:
- quieres eliminar duplicados,
- quieres comparar grupos,
- no te importa el orden de salida.

Vamos a empezar con un ejemplo muy simple.

````python
# EJEMPLO 1: convertir una lista con repetidos en un conjunto sin duplicados

nombres = ["Ana", "Juan", "Ana", "Carlos", "Luis", "Juan", "Sofía"]  # Creamos una lista; aquí sí se permiten repetidos.
print("Lista original", nombres)
print("tipo variable", type(nombres))
````

### Qué debes entender de este ejemplo

La línea más importante fue esta:

```python
nombres_unicos = set(nombres)
```

Aquí no estamos "borrando manualmente" los duplicados.  
Lo que hacemos es **cambiar de estructura de datos** para que la propia estructura nos ayude con el problema.

Ese es uno de los aprendizajes más importantes de Python:
> muchas veces el problema se resuelve mejor **eligiendo bien la estructura**, no escribiendo mucho código.

## 2. Ejemplo guiado: personas inscritas y personas que asistieron

Ahora veremos un caso parecido al tipo de razonamiento del taller, pero con datos distintos.

Tenemos:
- una lista de personas inscritas,
- una lista de personas que asistieron.

Primero observamos los datos.

````python
inscritos = ["Laura", "Mateo", "Sofía", "Laura", "Carlos", "Andrés", "Mateo", "Valentina"]  # Lista de inscritos; tiene repetidos.
asistentes = ["Mateo", "Carlos", "Sofía", "Valentina", "Sofía"]  # Lista de asistentes; también puede tener repetidos.

````

### Antes de programar, piensa

Hazte estas preguntas:
1. ¿Hay elementos repetidos?

## 3. `intersection()` y `&`: encontrar coincidencias

La intersección responde a preguntas como:
- ¿quiénes están en ambos grupos?
- ¿qué elementos coinciden?

Vamos a resolver la pregunta:
> ¿Quiénes estaban inscritos y además asistieron?

````python
coinciden = set_inscritos & set_asistentes
print("Personas que están en ambos grupos:", coinciden)
````

````python
coinciden_alt = set_incritos.intersection(set_asistentes)
print("Personas que están en ambos grupos:", coinciden_alt)
````

### Qué debes aprender aquí

Estas dos formas hacen lo mismo:

```python
set_inscritos.intersection(set_asistentes)
```

y

```python
set_inscritos & set_asistentes
```

La primera es más descriptiva.  
La segunda es más corta.

Ambas son válidas.

## 4. `difference()` y `-`: encontrar diferencias

Ahora vamos a responder esta pregunta:

> ¿Quiénes estaban inscritos pero no asistieron?

Eso es una **diferencia** entre conjuntos.

````python
solo_inscritos = set_incritos.difference(set_asistentes)
print("Personas inscritas que no asistieron:", solo_inscritos)
````

````python
solo_inscritos_alt = set_incritos - set_asistentes
print("Resultado usando -:", solo_inscritos_alt)
print("¿Ambas formas dan lo mismo?", solo_inscritos == solo_inscritos_alt)
````

### Idea clave

La dirección importa.

No es lo mismo:

```python
set_inscritos - set_asistentes
```

que

```python
set_asistentes - set_inscritos
```

La primera pregunta:
- ¿quién está en inscritos y no en asistentes?

La segunda pregunta:
- ¿quién está en asistentes y no en inscritos?

En conjuntos, **el orden del lado izquierdo y derecho sí cambia el significado**.

## 6. `for` + `if`: clasificar elementos uno por uno

Hasta ahora hemos comparado grupos completos.  
Ahora vamos a recorrer elemento por elemento para clasificarlos.

Pregunta:
> Si recorro todas las personas, ¿cómo sé si una estuvo en ambos grupos, solo en uno, o en otro caso?

Ahí usamos `for` y `if`.

### Qué significa la línea `if persona in coinciden`

- `persona` es el elemento actual del recorrido.
- `in` pregunta si ese elemento pertenece a un grupo.
- `coinciden` es el conjunto de los que aparecen en ambos grupos.

Entonces, esta línea se traduce así:

> "Si esta persona está dentro del grupo de coincidencias, clasifícala de esa manera."

## 8. `while`: recorrer con control manual

Ahora viene una lógica distinta.  
`for` recorre automáticamente.  
`while` recorre mientras una condición siga siendo verdadera.

Para usar `while` con datos de un conjunto, normalmente conviene convertir el conjunto en lista, porque así podemos usar posiciones con un índice.

### Qué debes vigilar en un `while`

En un `while`, hay tres piezas obligatorias:

1. **Una variable de control**  
   Ejemplo: `indice = 0`

2. **Una condición de continuación**  
   Ejemplo: `indice < len(lista_catalogo)`

3. **Una actualización de la variable**  
   Ejemplo: `indice += 1`

Si te falta la tercera, el ciclo puede quedarse ejecutándose sin fin.

## 9. `if` dentro de `while`: análisis más completo

Ahora no solo vamos a recorrer, sino también a clasificar mientras recorremos.

````python
lista_vendibles = list(vendibles)  # Convertimos a lista el conjunto de productos vendibles.
i = 0  # Inicializamos el contador para controlar el while.

while i < len(lista_vendibles):  # Repetimos mientras el contador siga dentro del rango válido.
    producto = lista_vendibles[i]  # Tomamos el producto actual según la posición i.

    if "Mouse" in producto:  # Evaluamos si la palabra Mouse aparece dentro del texto del producto actual.
        print(producto, "-> ejemplo de producto de entrada frecuente")  # Clasificación si cumple la condición.
    else:  # Caso contrario.
        print(producto, "-> producto disponible para despacho")  # Clasificación alternativa.

    i += 1  # Avanzamos al siguiente elemento para que el ciclo continúe correctamente.
````
