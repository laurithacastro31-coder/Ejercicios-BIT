2¿Qué género presentó el mayor número de casos?
Al analizar la distribución de pacientes por género, se observó que la categoría Female presentó el mayor número de casos registrados. Sin embargo, se identificaron las categorías "Unknown" y "Other" con una frecuencia considerable, lo que evidencia limitaciones en la calidad y completitud de los datos. Debido a esta situación, los resultados deben interpretarse con cautela, ya que la información faltante puede influir en la distribución real de los casos por género.

3¿Qué hospital presentó el mayor número de hospitalizaciones?
Se identificó que el HospitalID 75 registró la mayor cantidad de hospitalizaciones dentro del conjunto de datos analizado, con un total de 72 casos durante el periodo registrado en la base de datos.

4¿Cúal fue la edad con mas hospitalizaciones?
Al clasificar los pacientes por grupos etarios, se identificó que la categoría "Adulto Mayor" presentó el mayor número de hospitalizaciones. Este resultado sugiere que las personas de mayor edad fueron quienes requirieron con mayor frecuencia atención hospitalaria dentro de la población analizada, lo que puede estar relacionado con una mayor presencia de enfermedades crónicas y comorbilidades en este grupo poblacional asociado a la falta de controles pertinentes para cada grupo pobracional con seguimiento y prevfencion de enfermedades.



5¿Cúal fue la duracion promedio de la estancia hospitalaria de pacientes?
A partir del análisis de las fechas de ingreso y egreso, se calculó la duración de la estancia hospitalaria de los pacientes. Los resultados mostraron que el tiempo promedio de hospitalización fue de 5,15 días. Lo que indica que es una estancia prologada de lo habitual, esto con el fin de poder tomar decisiones frente a tiempos de incubacion de enfermedades asi como oportuno direccionamiento al servicio de urgencias


```python
import matplotlib.pyplot as plt

diagnosticos = {
    'CHOLELITHIASIS': 363,
    'URINARY TRACT INFECTION': 363,
    'OSTEOARTHRITIS': 352,
    'DIVERTICULITIS': 345,
    'ATRIAL FIBRILLATION': 345
}

plt.figure(figsize=(10,5))
plt.bar(diagnosticos.keys(), diagnosticos.values(), color='yellowgreen',edgecolor='black')

plt.title('Diagnósticos con mayor prevalencia')
plt.xlabel('Diagnóstico')
plt.ylabel('Número de casos')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
